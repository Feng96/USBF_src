# USBF: a real local example

Install the package and run `usbf fetch --data-dir data` from the repository root first. Run every cell in order in a fresh Python 3.11 kernel. This notebook performs actual B0/B1 inference, generates footprints, and re-scores fixed references. The three objects cover a changed outline, an unchanged building, and an unconfirmed circle candidate reverting to GBA. They are software-reproduction examples, not an additional validation sample.


```python
from pathlib import Path
import json, subprocess, sys
DATA_DIR = Path("../data").resolve()
OUTPUT_DIR = Path("../outputs/notebook").resolve()
def run(*args):
    result = subprocess.run([sys.executable, "-m", "usbf", *map(str, args)], check=True, text=True, capture_output=True)
    print(result.stdout)
run("verify", "--data-dir", DATA_DIR)
print("Kernel:", sys.executable)
```

```python
run("example", "--data-dir", DATA_DIR, "--output", OUTPUT_DIR, "--device", "cpu", "--threads", "2")
run("regression", "--data-dir", DATA_DIR, "--product", OUTPUT_DIR / "USBF.parquet", "--output", OUTPUT_DIR / "REGRESSION.json")
```

```python
import geopandas as gpd
product = gpd.read_parquet(OUTPUT_DIR / "USBF.parquet")
product[["key", "modified", "modification_type", "height_valid_fraction", "circle_gate_status", "circle_gate_reason"]]
```

```python
run("preview", "--data-dir", DATA_DIR, "--product", OUTPUT_DIR / "USBF.parquet", "--output", OUTPUT_DIR / "preview.png")
from IPython.display import Image, display
display(Image(filename=str(OUTPUT_DIR / "preview.png")))
```
